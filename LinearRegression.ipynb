# Linear Regression

```python
import numpy as np

from sklearn.datasets import load_diabetes, fetch_california_housing

data = load_diabetes()              # m≈442, n=10
X, y = data.data, data.target
```

```python
def normalize_features(X):
    """
    Standardize features:
      X_norm = (X - mean) / std
    Returns normalized X, means, stds.
    """
    mu    = np.mean(X, axis=0)
    sigma = np.std(X,  axis=0)
    X_norm = (X - mu) / sigma
    return X_norm, mu, sigma

def compute_cost(X, y, theta):
    """
    Compute Mean Squared Error cost for linear regression:
      J(theta) = (1/(2m)) * sum((X @ theta - y)^2)
    """
    m = len(y)
    preds = X.dot(theta)
    return (1/(2*m)) * np.sum((preds - y)**2)

def gradient_descent(X, y, theta, alpha, num_iters):
    """
    Perform gradient descent to learn theta.
    Returns final theta and history of cost values.
    """
    m = len(y)
    J_history = []
    for _ in range(num_iters):
        error    = X.dot(theta) - y
        gradient = (1/m) * X.T.dot(error)
        theta    = theta - alpha * gradient
        J_history.append(compute_cost(X, y, theta))
    return theta, J_history
```

```python
# Generate synthetic data
np.random.seed(0)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X[:,0] + np.random.randn(100)

# Feature scaling
X_scaled, mu, sigma = normalize_features(X)

# Add intercept term
X_reg = np.c_[ np.ones((100,1)), X_scaled ]

# Initialize parameters
theta_init = np.zeros(2)

# Run gradient descent
theta_lin, J_hist_lin = gradient_descent(X_reg, y, theta_init,
                                         alpha=0.1, num_iters=100)

print("Linear Regression learned parameters:", theta_lin)
print("Final cost (MSE):", J_hist_lin[-1])
```
